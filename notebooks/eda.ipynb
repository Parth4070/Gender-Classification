```python
from types import DynamicClassAttribute
from pathlib import Path

DATASET_DIR = Path("dataset")

train_dir = DATASET_DIR / "train"
test_dir = DATASET_DIR / "test"

print(train_dir)
print(test_dir)
```

```python
from pathlib import Path

extensions = {".jpg", ".jpeg", ".png"}

for split in ["train", "test"]:
    split_dir = DATASET_DIR / split

    print(f"\n{split.upper()}")

    for class_name in ["men", "women"]:
        class_dir = split_dir / class_name

        images = [
            p for p in class_dir.iterdir()
            if p.suffix.lower() in extensions
        ]

        print(f"{class_name}: {len(images)} images")
```

```python
import matplotlib.pyplot as plt
from PIL import Image
import random

men_dir = DATASET_DIR / "train" / "men"
women_dir = DATASET_DIR / "train" / "women"

men_images = list(men_dir.glob("*"))
women_images = list(women_dir.glob("*"))

fig, axes = plt.subplots(2, 5, figsize=(15, 6))

for i in range(5):
    image = Image.open(random.choice(men_images))

    axes[0, i].imshow(image)
    axes[0, i].set_title("Men")
    axes[0, i].axis("off")

for i in range(5):
    image = Image.open(random.choice(women_images))

    axes[1, i].imshow(image)
    axes[1, i].set_title("Women")
    axes[1, i].axis("off")

plt.tight_layout()
plt.show()
```
